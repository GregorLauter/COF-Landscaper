# COF-LZU-1 | Optional PXRD-guided scaling

Differences between simulated and experimental PXRD patterns can arise from the inherently idealised nature of periodic structural models. Framework defects, stacking disorder, layer buckling and guest molecules can influence the positions, shapes and intensities of experimental reflections. These effects and the associated challenges in interpreting PXRD patterns of 2D COFs are discussed in [The Devil Is in the Details](https://doi.org/10.1021/jacsau.4c00979), using COF-LZU-1 as a representative example.

Furthermore, geometry optimisation can lead to different local minima characterised by slightly different unit-cell parameters. Consequently, even when the main features of the experimental diffraction pattern are reproduced, systematic discrepancies in peak positions may remain.

In this example, we investigate whether these discrepancies can be reduced by adjusting the in-plane lattice parameters of the optimised serrated COF-LZU-1 structure, followed by geometry optimisation at fixed cell parameters.

**This functionality is experimental and is not part of the recommended default COF-Landscaper workflow.** Good agreement with experimental PXRD can often be obtained without any additional structural adjustment.

Lattice scaling should be approached with particular caution, as improved agreement with experimental PXRD does not necessarily imply a more accurate or physically meaningful structure. Increasingly large scaling corrections may introduce artificial structural distortions and indicate more fundamental discrepancies between the computational model and the experimental material. COF-Landscaper therefore issues warnings for scaling corrections exceeding 3%, with an additional warning above 10%. These thresholds are practical safeguards rather than validated criteria for structural accuracy.

The resulting structures should always be carefully inspected for distortions in bond lengths, bond angles and layer geometry. In-plane scaling modifies the lengths of the lattice vectors but does not independently correct discrepancies in cell angles or interlayer spacing.

The analysis and PXRD functions introduced in the main workflow are reused here. By specifying `source="postopt"`, COF-Landscaper automatically accesses the postoptimised structures and stores the corresponding results separately in `7_COF-LZU-1_postanalysis/`.

[Original COF-LZU-1 synthesis](https://doi.org/10.1021/ja206846p) | [Experimental data sources](../../DATA_SOURCES.md)

```python
import coflandscaper as cl
```

```python
COF_NAME = "COF-LZU-1"
MODE = "serr"
```

## Compare the original diffraction patterns

We first compare the simulated PXRD pattern of the optimised serrated structure with the experimental pattern. This allows us to identify systematic differences in peak positions before adjusting the in-plane lattice parameters.

```python
pxrd = cl.PXRD()
pxrd.run(COF_NAME, mode=MODE)
pxrd.plot_sim_vs_exp(COF_NAME, mode=MODE, xlim=(3.5, 30.0), save=False);
```

## Select reflections for lattice scaling

Select a few clearly identifiable reflections that correspond between the simulated and experimental PXRD patterns.

Only include reflections suitable for adjusting the in-plane lattice parameters. Reflections dominated by the interlayer distance should be excluded, as the interlayer spacing remains unchanged during scaling.

**Use this adjustment with caution.** Larger scaling corrections may indicate fundamental discrepancies between the structural model and the experimental material rather than a simple difference in lattice dimensions.

```python
peak_regions = [
    (4.0, 5.5),
    (7.5, 9.0),
]

pxrd.extract_peak_regions(COF_NAME, mode=MODE, peak_regions=peak_regions)
pxrd.generate_scaled_cif(COF_NAME, mode=MODE)
```

Carefully inspect the adjusted structure for distortions before proceeding with geometry optimisation. Improved agreement with experimental PXRD alone does not establish a physically meaningful structure.

The scaled structures are saved in `6_COF-LZU-1_scaling/scaling/`.

```python
cl.VisualizeCOF().view(f"{COF_NAME}/6_{COF_NAME}_scaling/scaling/{MODE}")
```

## Optimise the adjusted structure

Following the lattice adjustment, we optimise the atomic positions while keeping all unit-cell parameters fixed. This allows the local geometry to relax without altering the adjusted lattice dimensions.

The resulting structures are saved in `6_COF-LZU-1_scaling/postopt/serr/`.

```python
cl.GeometryOpt().run_fixed(COF_NAME, mode=MODE)
```

## Compare the diffraction patterns

We now compare the simulated PXRD pattern of the postoptimised structure with the experimental pattern to assess whether the lattice adjustment has improved the agreement.

**Better agreement with experimental PXRD does not necessarily imply a more accurate structural model.** Always inspect the resulting structure carefully to ensure that the lattice adjustment has not introduced unrealistic bond lengths, bond angles or distortions of the framework.

```python
post_pxrd = cl.PXRD()
post_pxrd.run(COF_NAME, mode=MODE, source="postopt")
post_pxrd.plot_sim_vs_exp(
    COF_NAME, mode=MODE, source="postopt", xlim=(3.5, 30.0), save=False
)
```
